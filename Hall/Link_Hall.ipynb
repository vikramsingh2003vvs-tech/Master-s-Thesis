# Link datasets magnetotrasporto cuprati (versione completa)

**Legenda stelle (qualità/completezza dataset per un dato materiale):**
- ★★★★★ : MR(H,T), R_H(H,T), R_H(T), ρ(T) e possibilità di costruire θ_H(T,B), cotθ_H(T) sullo stesso set di campioni, con intervalli di T e H ben sovrapposti.
- ★★★★☆ : dati molto completi, ma con qualche limitazione (es. θ_H non sempre riportato esplicitamente o range T/H meno esteso).
- ★★★☆☆ : buoni dati su ρ(T) e R_H(T), MR(H) o θ_H meno sistematici / distribuiti su più lavori.
- ★★☆☆☆ : dati utili ma frammentari per il tuo scopo (spesso solo R_H(T) e ρ(T)).

## Nd-LSCO (La1.6−xNd0.4SrxCuO4, p ≈ 0.20–0.24) – ★★★★★

### Nd-LSCO (p ≈ 0.20)
- **MR / ρ(B,T)**  
  - Daou et al., *Nature Phys.* 5, 31 (2009) – ρ(T) a 0, 15, 35 T, T da pochi K a ~80 K (MR piccola ma quantificabile):  
    <https://arxiv.org/pdf/0806.2881.pdf>  
  - Collignon et al., *Phys. Rev. B* 95, 224517 (2017) – ρ(H,T), MR(H,T) fino a 37.5 T, T criogeniche (pochi K → decine di K):  
    <https://arxiv.org/pdf/1607.05693.pdf>
- **R_H(T)**  
  - Daou et al. – R_H(T) per p = 0.20 con campi fino a ~30 T, T da pochi K a ~80 K:  
    <https://arxiv.org/pdf/0806.2881.pdf>  
  - Collignon et al. – R_H(T) per p = 0.20 sullo stesso set di campioni di MR(H,T):  
    <https://arxiv.org/pdf/1607.05693.pdf>
- **R_H(B)**  
  - Collignon et al. – isoterme R_H(H) per p = 0.20, T basse (~1–30 K) fino a 37.5 T:  
    <https://arxiv.org/pdf/1607.05693.pdf>
- **θ_H(T,B), cotθ_H(T)**  
  - Ricavabili combinando ρ(H,T) e R_H(H,T) di Collignon e ρ(T), R_H(T) di Daou (stessi campioni, T sovrapposte).

### Nd-LSCO (p ≈ 0.21)
- **MR / ρ(B,T)**  
  - Collignon et al. – ρ(H,T) isoterme da ~1.5 K a decine di K, fino a 37.5 T:  
    <https://arxiv.org/pdf/1607.05693.pdf>
- **R_H(T)**  
  - Collignon et al. – R_H(T) per p = 0.21, T da pochi K a ~80 K:  
    <https://arxiv.org/pdf/1607.05693.pdf>
- **R_H(B)**  
  - Collignon et al. – R_H(H) isoterme per p = 0.21, stesso intervallo T e campi di MR(H):  
    <https://arxiv.org/pdf/1607.05693.pdf>
- **θ_H(T,B), cotθ_H(T)**  
  - Direttamente costruibili da ρ(H,T) e R_H(H,T) nello stesso paper (intervalli T e H coincidenti).

### Nd-LSCO (p ≈ 0.22)
- **MR / ρ(B,T)**  
  - Collignon et al. – ρ(H,T), MR(H,T) dettagliata (figure 2–3–9), T basse (≈1–30 K), H fino a 37.5 T:  
    <https://arxiv.org/pdf/1607.05693.pdf>
- **R_H(T)**  
  - Collignon et al. – R_H(T) per p = 0.22, stesse T di ρ(T):  
    <https://arxiv.org/pdf/1607.05693.pdf>
- **R_H(B)**  
  - Collignon et al. – R_H(H) isoterme per p = 0.22, T ≈1–30 K:  
    <https://arxiv.org/pdf/1607.05693.pdf>
- **θ_H(T,B), cotθ_H(T)**  
  - Da ρ(H,T) e R_H(H,T) dello stesso lavoro, T e H perfettamente confrontabili.

### Nd-LSCO (p ≈ 0.23)
- **MR / ρ(B,T)**  
  - Collignon et al. – ρ(H,T) e MR(H,T) per p = 0.23 (1–30 K, fino a 37.5 T):  
    <https://arxiv.org/pdf/1607.05693.pdf>
- **R_H(T)**  
  - Collignon et al. – R_H(T) per p = 0.23, stesso range T:  
    <https://arxiv.org/pdf/1607.05693.pdf>
- **R_H(B)**  
  - Collignon et al. – R_H(H) per p = 0.23 (stessi T e H di MR):  
    <https://arxiv.org/pdf/1607.05693.pdf>
- **θ_H(T,B), cotθ_H(T)**  
  - Da ρ(H,T) e R_H(H,T), buono per confrontare upturn di ρ(T) e comportamento più regolare di R_H(T).

### Nd-LSCO (p ≈ 0.24)
- **MR / ρ(B,T)**  
  - Collignon et al. – ρ(H,T), MR(H,T) per p = 0.24 (T basse → decine di K, fino a 37.5 T):  
    <https://arxiv.org/pdf/1607.05693.pdf>  
  - Daou et al. – ρ(T) a 0, 15, 35 T (T fino a decine di K), MR molto piccola:  
    <https://arxiv.org/pdf/0806.2881.pdf>
- **R_H(T)**  
  - Daou et al. – R_H(T) per p = 0.24 (H ≈ 16 T), T da pochi K a ~80 K:  
    <https://arxiv.org/pdf/0806.2881.pdf>  
  - Collignon et al. – R_H(T) per p = 0.24:  
    <https://arxiv.org/pdf/1607.05693.pdf>
- **R_H(B)**  
  - Per p = 0.24 R_H(H) esplicito è meno dettagliato; si usano soprattutto R_H(T) a campo fisso nei due lavori.
- **θ_H(T,B), cotθ_H(T)**  
  - Da ρ(T), ρ(H,T) e R_H(T) combinati (Daou + Collignon), T da pochi K a ~80 K.

## LSCO (La2−xSrxCuO4, x ≈ 0.08–0.25) – ★★★☆☆

- **MR / ρ(B,T)**  
  - Boebinger et al., “Insulator-to-metal crossover in La2−xSrxCuO4 near optimum doping” – ρ(T,B) fino a 61 T per vari x, T da pochi K a ~100 K (soprattutto under/ottimo doping).
- **R_H(T)**  
  - Ando et al., “Evolution of the Hall coefficient and the peculiar electronic structure of La2−xSrxCuO4” – R_H(T) per x = 0.02–0.25, T da ~5 K a ~300 K a campo fisso:  
    <https://harvest.aps.org/v2/journals/articles/10.1103/PhysRevLett.92.197001/fulltext>
- **R_H(B)**  
  - Balakirev et al., “Fermi surface reconstruction at optimum doping in high-Tc superconductors” – R_H(H) ad alto campo vicino al doping ottimale (T poche decine di K).
- **θ_H(T), cotθ_H(T)**  
  - Da costruire combinando ρ(T) (serie di Ando su LSCO e altri lavori di resistività) con R_H(T) di Ando 2004; per molti x si osserva cotθ_H ~ T² su un ampio range di T.

## Tl2Ba2CuO6+δ (Tl2201, overdoped) – ★★★★☆

- **MR / ρ(B,T)**  
  - Mackenzie et al., “Normal-state magnetotransport in superconducting Tl2Ba2CuO6+δ to millikelvin temperatures” – ρ(T,B) da mK fino a decine di K, campi multi-tesla:  
    <https://pubmed.ncbi.nlm.nih.gov/9984193/>  
  - Hussey et al., “Tracking anisotropic scattering in overdoped Tl2Ba2CuO6+δ above 100 K” – ADMR per T fino a ~100 K:  
    <https://www.arxiv.org/abs/0905.2504>
- **R_H(T)**  
  - Mackenzie et al. – R_H(T) per diversi livelli di overdoping (δ variabile), T da mK a decine di K:  
    <https://pubmed.ncbi.nlm.nih.gov/9984193/>
- **R_H(B)**  
  - Mackenzie et al. – R_H(H) a varie T nel medesimo lavoro (campi multi-tesla, T molto basse):  
    <https://pubmed.ncbi.nlm.nih.gov/9984193/>
- **θ_H(T,B), cotθ_H(T)**  
  - Ricavabili da ρ(T,B) e R_H(T,B); l’analisi della separazione dei canali di scattering isotropo/anisotropo è discussa in Hussey et al.

## LCCO (La2−xCexCuO4, x = 0.08–0.17) – ★★★★★

- **MR / ρ(B,T)**  
  - Poniatowski/Sarkar/Greene, “Anomalous normal-state magnetotransport in an electron-doped cuprate La2−xCexCuO4” (*Phys. Rev. B* 103, 125102) – MR(H,T) con dipendenza ~H^{3/2} per T > 100 K su tutta la cupola superconduttiva, T tipicamente da ~100 K fino a qualche centinaio di K:  
    <https://par.nsf.gov/servlets/purl/10246739>  
    <https://www.emergentmind.com/papers/1912.00974>  
  - “Strange Metallic Transport in the Antiferromagnetic Regime of Electron Doped Cuprates” – MR e Hall su film LCCO per T da 0.7 a 45 K e campi fino a 65 T (H-lineare, scaling H–T):  
    <https://www.emergentmind.com/papers/2007.12765>
- **R_H(T)**  
  - Stesso lavoro Poniatowski/Sarkar – R_H(T) per x = 0.08–0.17, T fino a ~400 K; resistività mostra ρ ~ T² nel range 100–400 K:  
    <https://par.nsf.gov/servlets/purl/10246739>
- **R_H(B)**  
  - Poniatowski/Sarkar – R_H(H) a varie T e x, nello stesso range di H delle misure di MR(H,T):  
    <https://par.nsf.gov/servlets/purl/10246739>  
  - Strange Metal – dipendenza anomala di R_H con il campo sotto 50 K:  
    <https://www.emergentmind.com/papers/2007.12765>
- **θ_H(T,B), cotθ_H(T)**  
  - Poniatowski/Sarkar – cotθ_H(T) esplicita: ~T⁴ per x < x_FSR e ~T^{2.5} per x > x_FSR (x_FSR ≈ 0.14), per T > 100 K:  
    <https://par.nsf.gov/servlets/purl/10246739>

## YBCO (YBa2Cu3O7−δ, vari p) – ★★★☆☆

- **MR / ρ(B,T)**  
  - Diversi lavori di Boebinger, LeBoeuf, Doiron-Leyraud, ecc., su MR ab-plane ad alto campo (fino a decine di tesla, T da pochi K a ~100 K); spesso centrati su ricostruzione di Fermi surface.
- **R_H(T)**  
  - Lavori su pseudogap e CDW in YBCO riportano R_H(T) ad alto campo per vari dopaggi, T da pochi K a ~100–150 K.
- **R_H(B)**  
  - In diversi articoli sulla ricostruzione della Fermi surface in YBCO compaiono R_H(H) isoterme (T ~1–50 K, campi fino a ~60 T).
- **θ_H(ω,T), cotθ_H(T)**  
  - Grayson et al., “Spectral measurement of the Hall angle response in normal state cuprate superconductors” – misura θ_H(ω,T) (da dc a FIR) su film di YBCO, T tipicamente da poche decine a ~300 K:  
    <https://arxiv.org/abs/cond-mat/0108553>  
    <https://physics.umd.edu/~schmadel/Publications/Phys.Rev.Lett.89,037003(2002).pdf>

## Bi-based cuprati (Bi2Sr2CuO6+x, Bi2212, Bi2201, …) – ★★☆☆☆

- **MR / ρ(B,T)**  
  - “Aspects of normal-state resistivity of the cuprate superconductors Bi2Sr2CuO6+x, Tl2Ba2CuO6+x, and HgBa2CuO4+x” – resistività normale con T fino a ~300 K; MR(H) discussa ma non sempre sistematica:  
    <https://physics.ucsc.edu/~sriram/papers/paper_175.pdf>
- **R_H(T)**  
  - Vari articoli su R_H(T) e n_H in Bi2212/Bi2201 (spesso fino a ~300 K) riportati nelle review sui cuprati.
- **R_H(B)**  
  - Meno frequenti; in molti casi è riportato solo R_H(T) a campo fisso (pochi tesla).
- **θ_H(T), cotθ_H(T)**  
  - Classici risultati che mostrano cotθ_H ~ T² in Bi-based, distribuiti su più articoli; utili come confronto qualitativo ma non come dataset unico e completo.

## Riepilogo: materiali 4★/5★ consigliati per fit completi

- **Nd-LSCO (p = 0.20–0.24)** – ideale per confrontare pseudogap vs overdoped in un unico sistema, con dataset ρ(H,T), R_H(T,H) molto ben allineati.  
- **LCCO (x = 0.08–0.17)** – permette di testare dipendenze non banali di cotθ_H(T) (T⁴, T^{2.5}) e scaling H–T in MR.  
- **Tl2201 (overdoped)** – caso Fermi-liquido pulito, utile come “benchmark” con scattering anisotropo ben caratterizzato.
