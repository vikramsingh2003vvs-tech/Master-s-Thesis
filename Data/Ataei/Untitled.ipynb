```python
import pandas as pd

# 1. Carichiamo i tre file (saltando le prime due righe di commento con skiprows=2)
# Assicurati che i nomi dei file corrispondano a quelli nella tua cartella

# Carica Nd-LSCO (Temperature 4K, 40K, 80K)
moesm2 = pd.read_csv('41567_2022_1763_MOESM2_ESM.txt', sep='\t', header=None, skiprows=2, 
                     names=['B_4K', 'rho_4K', 'B_40K', 'rho_40K', 'B_80K', 'rho_80K'])

# Carica LSCO S1 (Temperature 4K, 40K, 80K)
moesm4 = pd.read_csv('41567_2022_1763_MOESM4_ESM.txt', sep='\t', header=None, skiprows=2, 
                     names=['B_4K', 'rho_4K', 'B_40K', 'rho_40K', 'B_80K', 'rho_80K'])

# Carica il file a 30K per entrambi
moesm6 = pd.read_csv('41567_2022_1763_MOESM6_ESM.txt', sep='\t', header=None, skiprows=2, 
                     names=['B_S1_30K', 'rho_S1_30K', 'B_S2_30K', 'rho_S2_30K', 'B_Nd_30K', 'rho_Nd_30K'])

# 2. Creiamo i dataset finali unendo le colonne di MOESM6 ai rispettivi file
# Uniamo Nd-LSCO (MOESM2 + le ultime due colonne di MOESM6)
nd_lsco_completo = pd.concat([moesm2, moesm6[['B_Nd_30K', 'rho_Nd_30K']]], axis=1)

# Uniamo LSCO S1 (MOESM4 + le prime due colonne di MOESM6)
lsco_s1_completo = pd.concat([moesm4, moesm6[['B_S1_30K', 'rho_S1_30K']]], axis=1)

# 3. Mostriamo l'anteprima delle tabelle unite
print("=== Anteprima Nd-LSCO Completo ===")
display(nd_lsco_completo.head())

print("\n=== Anteprima LSCO S1 Completo ===")
display(lsco_s1_completo.head())

# (Opzionale) De-commenta queste righe per salvare i nuovi file sul tuo computer
# nd_lsco_completo.to_csv('Nd_LSCO_Tutte_Temperature.csv', index=False)
# lsco_s1_completo.to_csv('LSCO_S1_Tutte_Temperature.csv', index=False)
```

```python
import pandas as pd

# 1. Carichiamo i tre file
moesm2 = pd.read_csv('41567_2022_1763_MOESM2_ESM.txt', sep='\t', header=None, skiprows=2, 
                     names=['B_4K', 'rho_4K', 'B_40K', 'rho_40K', 'B_80K', 'rho_80K'])

moesm4 = pd.read_csv('41567_2022_1763_MOESM4_ESM.txt', sep='\t', header=None, skiprows=2, 
                     names=['B_4K', 'rho_4K', 'B_40K', 'rho_40K', 'B_80K', 'rho_80K'])

moesm6 = pd.read_csv('41567_2022_1763_MOESM6_ESM.txt', sep='\t', header=None, skiprows=2, 
                     names=['B_S1_30K', 'rho_S1_30K', 'B_S2_30K', 'rho_S2_30K', 'B_Nd_30K', 'rho_Nd_30K'])

# 2. Uniamo le colonne
nd_lsco_completo = pd.concat([moesm2, moesm6[['B_Nd_30K', 'rho_Nd_30K']]], axis=1)
lsco_s1_completo = pd.concat([moesm4, moesm6[['B_S1_30K', 'rho_S1_30K']]], axis=1)

# 3. Salviamo in formato .txt (usando la tabulazione come separatore)
nd_lsco_completo.to_csv('Nd_LSCO_Completo.txt', sep='\t', index=False)
lsco_s1_completo.to_csv('LSCO_S1_Completo.txt', sep='\t', index=False)

print("File .txt salvati con successo nella cartella di Jupyter!")
```
