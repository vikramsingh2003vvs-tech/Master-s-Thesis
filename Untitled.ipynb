```python
import matplotlib.pyplot as plt
import numpy as np
from ipywidgets import interact

def plot_completo(frequenza, ampiezza):
    x = np.linspace(0, 2 * np.pi, 1000)
    # 1. Moltiplichiamo per l'ampiezza
    y = ampiezza * np.sin(frequenza * x)
    
    plt.figure(figsize=(8, 5))
    plt.plot(x, y, color='blue', lw=2)
    plt.title(f'Seno: Freq={frequenza}, Amp={ampiezza}')
    plt.grid(True)
    
    # 2. Adattiamo i limiti della Y all'ampiezza massima per non tagliare il grafico
    plt.ylim(-2.1, 2.1) 
    plt.show()

# 3. Chiamata corretta (seno l'argomento 'update')
interact(plot_completo, frequenza=(1, 10, 0.5), ampiezza=(0.1, 2.0, 0.1));
```
